### Ejercicio I1: Conversor de temperatura

Pide una temperatura en grados Celsius (float) y muestra su equivalente en Fahrenheit (`F = C * 9/5 + 32`) y en Kelvin (`K = C + 273.15`), cada uno redondeado a 1 decimal con formato de f-string (`:.1f`).

```python
print ("Conversor de Temperatura")
print ("Dame una temperatura en Celsius?")
tempC = float(input())

tempF = tempC * 9/5 +32
tempK = tempC + 273.15

print ("La temperatura es:")
print(f"Celsius: {tempC:.1f}Cº")
print(f"Fahrenheit: {tempF:.1f}Fº")
print(f"Kelvin: {tempK:.1f}Kº")
```

### Ejercicio I3: Comprobaciones con operadores de comparacion

Declara dos numeros `a` y `b`, y muestra en pantalla el resultado (`True`/`False`) de compararlos con `==`, `!=`, `>`, `<`, `>=` y `<=`, uno por linea.

```python
print ("Comparador de numeros")
print ("Cual es el primer numero?:")
a = float(input())
print ("Cual es el segundo numero?:")
b = float(input())

print("Son iguales?: ", a == b)
print("No son iguales?: ", a != b)
print("El primer numero es mayor al segundo: ", a > b)
print("El primer numero es menor al segundo: ", a < b)
print("El primer numero es mayor o igual al segundo: ", a >= b)
print("El primer numero es menor o igual al segundo: ", a <= b)
```

### Ejercicio A2: Depuracion en la practica

El siguiente codigo tiene un error. Ejecutalo, lee el mensaje de error con calma, y corrigelo usando lo aprendido sobre depuracion (no lo reescribas entero: identifica la linea exacta del fallo).

```python
def calcular_media(notas):
    total = 0
    for nota in notas:
        total = total + nota
    media = total / len(notas)  # <- el error estaba en que no se estaba calculando la longitud de la lista NOTAS, sino la de un numero que formaba parte de la lista.
    return media

print(calcular_media([7, 8, 9, 6]))
```
