### Ejercicio B1: Funcion es_primo

Escribe una funcion `es_primo(n)` que devuelva `True` si `n` es un numero primo y `False` en caso contrario. Pruebala con varios numeros.

```python
num = int(input("Dame un numero: "))

contador = 0

for divisor in range(1, num):
    if num % divisor == 0:
        contador +=1

if contador <= 2:
    print("El numero es primo")
else: 
    print("El numero no es primo")
```

### Ejercicio I1: Estadisticas de una lista

Escribe una funcion `estadisticas(numeros)` que reciba una lista y devuelva una tupla `(minimo, maximo, media)`, usando retorno multiple.

```python
import math

numeros = [1, 34, 56, 72, 85, 97, 1016, 1995, 543, 1, 0, 16, 22]

def estadisticas(numeros):
    minimo = min(numeros)
    maximo = max(numeros)
    media = sum(numeros)/len(numeros)
    
    return f"Min: {minimo} \nMax: {maximo} \nMedia:  {media}"

print(estadisticas(numeros))
```

### Ejercicio A1: Ordenar con lambda

Dada una lista de diccionarios de estudiantes (con `nombre` y `nota`), ordenala de mayor a menor nota usando `sorted()` con `key=lambda` (sin definir una funcion aparte).

```python
alumnos = [("Ana", 7), ("Luis", 9), ("Marta", 5), ("Pedro", 9)]

print(sorted(alumnos, key=lambda alumno: alumno[1]))
```

### Ejercicio A2: Decorador contador de llamadas

Escribe un decorador `@contar_llamadas` que lleve la cuenta de cuantas veces se ha llamado a la funcion decorada, e imprima esa cuenta cada vez.

```python
def contar_llamadas(funcion):
    contador = 0

    def envoltura():
        nonlocal contador
        contador += 1
        resultado = funcion()
        print("Se ha llamado:", contador)
        return resultado

    return envoltura


@contar_llamadas
def preguntar():
    respuesta = input("¿Quieres seguir? (s/n): ").strip().lower()
    return respuesta == "s"


while preguntar():
    print("¡Seguimos!")

print("Fin del programa")
```
