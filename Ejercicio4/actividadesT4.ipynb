### Ejercicio B1: Entrada numerica segura

Pide un numero por teclado dentro de un `try`/`except`, capturando `ValueError` si el usuario escribe algo que no es un numero, y muestra un mensaje de error en ese caso.

```python
try:
    numero = int(input("Introduce un número: "))
    print(f"Has introducido el número {numero}")
except ValueError:
    print("Error: lo que has escrito no es un número válido.")
```

### Ejercicio B2: Division segura

Escribe una funcion `dividir_seguro(a, b)` que devuelva `a / b`, capturando `ZeroDivisionError` y devolviendo `None` con un mensaje si `b` es 0.

```python
def operacion_completa(a, b):
    try:
        resultado = a / b
    except ZeroDivisionError:
        print("ERROR: División por cero")
        resultado = None
    except TypeError:
        print("ERROR: Tipos de datos incorrectos")
        resultado = None
    else:
        print(f"Operación exitosa: resultado = {resultado}")
    finally:
        print("Operación finalizada")
    return resultado

a = int(input("Introduce un número: "))
b = int(input("Introduce un número: "))

operacion_completa(a,b)
```

### Ejercicio I1: Validar edad con raise

Escribe una funcion `validar_edad(edad)` que lance `ValueError` si la edad es negativa o mayor de 120, y la use dentro de un `try`/`except` que capture ese error y muestre el mensaje.

```python
def validar_edad(edad):
    if edad < 0:
        raise ValueError("La edad no puede ser negativa")
    if edad > 120:
        raise ValueError("La edad no es realista")
    if not isinstance(edad, int):
        raise TypeError("La edad debe ser un número entero")
    return True

edad = int(input("Introduce tu edad: "))

try:
    validar_edad(edad)
except ValueError as e:
    print(f"Error de valor: {e}")
```

### Ejercicio I2: Pedir hasta que sea valido

Con un bucle `while` combinado con `try`/`except`, sigue pidiendo un numero entero hasta que el usuario introduzca uno valido.

```python
while True:
    try:
        numero = int(input("Introduce un número entero: "))
        break
    except ValueError:
        print("Error: debes escribir un número entero. Inténtalo de nuevo.")

print(f"Has introducido el número {numero}")
```
