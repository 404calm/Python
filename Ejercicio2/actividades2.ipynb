### Ejercicio B3: Adivina la contrasenha

Con un bucle `while`, pide una contrasenha repetidamente hasta que el usuario escriba exactamente `"python123"`. Al acertar, muestra un mensaje de exito.

```python
contrasena = str(input("Introduce la contraseña: "))

while contrasena != str("python123"):
    print("La contraseña es incorrecta")
    contrasena = str(input("Introduce la contraseña: "))
    
print("La contraseña es correcta")
```

### Ejercicio I1: Clasificador de notas

Pide una nota entre 0 y 10 y muestra la calificacion con `if`/`elif`/`else`: Sobresaliente (>=9), Notable (>=7), Bien (>=6), Suficiente (>=5), Insuficiente (resto).

```python
nota = float(input("¿Que nota quieres clasificar?: "))

if nota <= 10 and nota >= 9:
    print("Tienes un sobresaliente.")
elif nota < 9 and nota >= 7:
    print("Tienes un notable.")
elif nota < 7 and nota >= 6:
    print("Tienes un bien.")
elif nota < 6 and nota >= 5:
    print("Tienes un suficiente.")
elif nota < 5 and nota >= 0:
    print("Tienes un suspenso.")
else: 
    print("La nota no se encuentra entr 0 y 10.")
```

### Ejercicio I2: FizzBuzz

Recorre los numeros del 1 al 30. Si es multiplo de 3, imprime `Fizz`; si es multiplo de 5, imprime `Buzz`; si es multiplo de ambos, imprime `FizzBuzz`; si no, imprime el numero.

```python
suma = 0
for num in range(1,1001):
    if num % 3 == 0 or num % 5 == 0:
        suma += num
print(suma)    
```

### Ejercicio A1: Codigo de estado HTTP con match

Dado un codigo de estado HTTP (como entero: 200, 301, 404, 500 u otro), usa `match`/`case` para imprimir su categoria: 'Exito' (200), 'Redireccion' (301), 'No encontrado' (404), 'Error del servidor' (500), o 'Desconocido' para cualquier otro.

```python
codigoHttp = int(input("¿Cual es el codigo http?: "))
match codigoHttp:
    case 200:
        print("Exito' (200)")
    case 301:
        print("Redireccion' (301)")
    case 404: 
        print("No encontrado' (404)")
    case 500: 
        print("Error del servidor' (500)")
    case _: 
        print("Desconocido' para cualquier otro")
```

### Ejercicio A2: Semaforo con match y bucle controlado

Simula un semaforo que cambia de color 6 veces (rojo -> verde -> ambar -> rojo...) usando un bucle `for` con un contador y `match`/`case` para decidir el siguiente color a partir del actual.

```python
color = str("")

for contador in range(1,7):
    match color:
        case str("verde"):
            print("ambar")
            color = "ambar"
        case str("ambar"):
            print("rojo")
            color = "rojo"
        case str("rojo"):
            print("verde")
            color = "verde"
        case _:
            print("rojo")
            color = "rojo"
```
